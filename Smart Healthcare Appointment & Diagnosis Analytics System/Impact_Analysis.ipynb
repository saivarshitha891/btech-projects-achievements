# 🧠 Smart Healthcare Appointment & Diagnosis Analytics – Impact & Value Showcase

This impact analysis outlines how the system delivered measurable improvements by cleaning, transforming, and analyzing simulated healthcare data using **Python (Pandas)** and **SQL concepts**.

✅ This proves real-world data skills such as:
- Handling missing or messy records
- Detecting duplicates and anomalies
- Analyzing cancellations, revenue trends
- Supporting data-driven decisions for healthcare efficiency


```python
import random
import numpy as np

# Predefined name parts
first_names = ['Aarav', 'Vivaan', 'Aditya', 'Rohan', 'Diya', 'Ishita', 'Meera', 'Tanvi', 'Pooja', 'Saanvi']
last_names = ['Sharma', 'Patel', 'Reddy', 'Mehta', 'Joshi', 'Nair', 'Desai', 'Verma', 'Iyer', 'Kapoor']

def generate_fake_name():
    return random.choice(first_names) + " " + random.choice(last_names)

def generate_fake_phone():
    return f"+91-9{random.randint(100000000, 999999999)}"
```

```python
import pandas as pd

num_patients = 1500
patients = pd.DataFrame({
    'patient_id': range(1, num_patients + 1),
    'name': [generate_fake_name() for _ in range(num_patients)],
    'age': np.random.randint(20, 70, size=num_patients),
    'gender': np.random.choice(['Male', 'Female'], size=num_patients),
    'contact': [generate_fake_phone() for _ in range(num_patients)]
})

print(patients)
```

```python
specializations = ['Cardiology', 'Dermatology', 'Neurology', 'Pediatrics', 'Orthopedics']

num_doctors = 60
doctors = pd.DataFrame({
    'doctor_id': range(1, num_doctors + 1),
    'name': [f"Dr. {generate_fake_name()}" for _ in range(num_doctors)],
    'specialization': np.random.choice(specializations, size=num_doctors),
    'availability': np.random.choice([True, False], size=num_doctors)
})

doctors.head()
```

```python
# 3. Generate Appointments
num_appointments = 3000
appointments = pd.DataFrame({
    'appointment_id': range(1, num_appointments + 1),
    'patient_id': np.random.choice(patients['patient_id'], num_appointments),
    'doctor_id': np.random.choice(doctors['doctor_id'], num_appointments),
    'appointment_date': pd.to_datetime(np.random.choice(pd.date_range('2024-12-01', '2025-03-31'), size=num_appointments)),
    'status': np.random.choice(['Completed', 'Cancelled'], size=num_appointments)
})

print(appointments)
```

```python
# 4. Diagnoses
diagnoses = pd.DataFrame({
    'diagnosis_id': range(1, num_appointments + 1),
    'appointment_id': appointments['appointment_id'],
    'symptoms': np.random.choice(['Fever', 'Cough', 'Chest Pain', 'Skin Rash', 'Headache'], size=num_appointments),
    'diagnosis': np.random.choice(['Flu', 'Allergy', 'Infection', 'Hypertension'], size=num_appointments),
    'prescription': np.random.choice(['Paracetamol', 'Antibiotic', 'Antihistamine', 'Painkiller'], size=num_appointments)
})

print(diagnoses)
```

```python
# 5. Billing
billing = pd.DataFrame({
    'bill_id': range(1, num_appointments + 1),
    'appointment_id': appointments['appointment_id'],
    'amount': np.round(np.random.uniform(150.0, 1200.0, size=num_appointments), 2),
    'payment_status': np.random.choice(['Paid', 'Pending'], size=num_appointments)
})

print(billing)
```

```python
### 📅 Insight 1: Duplicate records

# Check duplicate patients based on name and contact
duplicates = patients.duplicated(subset=['name', 'contact'], keep=False)
duplicate_records = patients[duplicates]
print(f"🔍 Found {duplicate_records.shape[0]} duplicate patient entries")
```

```python
### 📅 Insight 2: Missed or Canceled Appointments
# Calculated cancellation rate to measure inefficiencies in scheduling.


total_appointments = appointments.shape[0]
cancelled = appointments[appointments['status'] == 'Cancelled'].shape[0]
cancelled_percent = (cancelled / total_appointments) * 100
print(f"📉 Total Appointments: {total_appointments}")
print(f"❌ Canceled Appointments: {cancelled} ({cancelled_percent:.2f}%)")
```

```python
### 💰 Insight 3: Revenue Loss from Canceled Appointments
# Estimated monthly revenue loss from canceled appointments using billing data.

canceled_ids = appointments[appointments['status'] == 'Cancelled']['appointment_id']
lost_revenue = billing[billing['appointment_id'].isin(canceled_ids)]['amount'].sum()
print(f"💸 Estimated monthly revenue lost due to cancellations: ₹{lost_revenue}")
```

```python
### 📉 Insight 4: Billing Anomaly Detection
# Detected potential fraud or incorrect entries in billing using IQR outlier method.

Q1 = billing['amount'].quantile(0.25)
Q3 = billing['amount'].quantile(0.75)
IQR = Q3 - Q1
outliers = billing[(billing['amount'] < Q1 - 1.5 * IQR) | (billing['amount'] > Q3 + 1.5 * IQR)]
print(f"🚨 Billing outliers detected: {len(outliers)}")
```

```python
# Add date column to the appointments dataframe
import numpy as np
import pandas as pd

# Simulate appointment dates: last 6 months
date_range = pd.date_range(start='2024-11-01', end='2025-04-30', freq='D')

# Randomly assign a date to each appointment
np.random.seed(42)  # reproducible results
appointments['date'] = np.random.choice(date_range, size=len(appointments))

# Check new column
appointments[['appointment_id', 'date']].head()
```

```python
# Merge billing and appointments to access 'date'
monthly_data = billing.merge(appointments[['appointment_id', 'date']], on='appointment_id', how='left')

# Ensure datetime format
monthly_data['date'] = pd.to_datetime(monthly_data['date'], errors='coerce')

# Create month column
monthly_data['month'] = monthly_data['date'].dt.to_period('M')

# Monthly revenue
monthly_revenue_trend = monthly_data.groupby('month')['amount'].sum().reset_index()
monthly_revenue_trend['month'] = monthly_revenue_trend['month'].astype(str)

# Display
monthly_revenue_trend
```

```python
# Revenue Trend Summary + % Increase
# Calculate revenue per month
monthly_revenue_trend = monthly_data.groupby('month')['amount'].sum().reset_index()
monthly_revenue_trend['month'] = monthly_revenue_trend['month'].astype(str)

# Calculate % change between consecutive months
monthly_revenue_trend['pct_change'] = monthly_revenue_trend['amount'].pct_change().fillna(0) * 100
monthly_revenue_trend['pct_change'] = monthly_revenue_trend['pct_change'].round(2)

# Rename for display
monthly_revenue_trend.rename(columns={
    'month': 'Month',
    'amount': 'Revenue (₹)',
    'pct_change': 'Monthly Change (%)'
}, inplace=True)

# Overall % increase: last month vs first month
first = monthly_revenue_trend.iloc[0]['Revenue (₹)']
last = monthly_revenue_trend.iloc[-1]['Revenue (₹)']
overall_change = ((last - first) / first) * 100 if first != 0 else 0
overall_change = round(overall_change, 2)

# Display table and summary
display(monthly_revenue_trend)
print(f"📈 Overall Revenue Increase from {monthly_revenue_trend.iloc[0]['Month']} to {monthly_revenue_trend.iloc[-1]['Month']}: {overall_change}%")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(monthly_revenue_trend['Month'], monthly_revenue_trend['Revenue (₹)'], marker='o', label='Revenue')
plt.bar(monthly_revenue_trend['Month'], monthly_revenue_trend['Change (%)'], alpha=0.4, label='% Change')
plt.xticks(rotation=45)
plt.title("Monthly Revenue & % Trend")
plt.xlabel("Month")
plt.ylabel("Amount / % Change")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
